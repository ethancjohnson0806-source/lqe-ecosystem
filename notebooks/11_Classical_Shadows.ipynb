# 11 — Classical Shadows

**Concept.** Random Pauli measurements build a classical snapshot for many-property estimation.

```python
import os, sys
_LQE = os.environ.get("LQE_PATH", os.path.abspath(os.path.join("..", "Legitimate-Quantum-Engine")))
if _LQE not in sys.path: sys.path.insert(0, _LQE)
from legitimate_quantum_engine.core.statevector import StatevectorSim
from legitimate_quantum_engine.characterization.classical_shadows import ClassicalShadow

sim = StatevectorSim(2)
sim.apply('H', 0); sim.apply('CNOT', 0, 1)
shadow = ClassicalShadow(num_qubits=2, num_snapshots=80)
shadow.random_pauli_measurement(sim)
print(f'Snapshots: {len(shadow.snapshots)}')
for obs in ['ZZ', 'XX']:
    print(f'  <{obs}> ≈ {float(shadow.estimate_expectation(obs)):.4f}')
```

**Takeaway.** Shadows predict many observables without re-preparing the state for each.
