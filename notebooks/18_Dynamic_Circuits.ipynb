# 18 — Dynamic Circuits

**Concept.** Mid-circuit measurement + feedforward enables teleportation.

```python
import os, sys
_LQE = os.environ.get("LQE_PATH", os.path.abspath(os.path.join("..", "legitimate_quantum_engine_v5.0")))
if _LQE not in sys.path: sys.path.insert(0, _LQE)
from legitimate_quantum_engine.circuit_tools.dynamic_circuits import DynamicStatevectorSim
import numpy as np

sim = DynamicStatevectorSim(3)
sim.apply('H', 0); sim.apply('H', 1); sim.apply('CNOT', 1, 2)
sim.apply('CNOT', 0, 1); sim.apply('H', 0)
m0, m1 = sim.measure(0), sim.measure(1)
if m1 == 1: sim.conditional_apply(1, 2, 'X', 2)
if m0 == 1: sim.conditional_apply(1, 2, 'Z', 2)
print('bits', m0, m1, 'probs', np.round(sim.probabilities(), 4))
```

**Takeaway.** Feedforward is a building block for fault-tolerant control.
