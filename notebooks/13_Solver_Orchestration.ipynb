# 13 — Solver Orchestration

**Concept.** An orchestrator picks solvers from problem size and structure.

```python
import os, sys
_LQE = os.environ.get("LQE_PATH", os.path.abspath(os.path.join("..", "legitimate_quantum_engine_v5.0")))
_ECO = os.environ.get("LQE_ECOSYSTEM_PATH", os.path.abspath(".."))
for _p in (_LQE, _ECO):
    if _p not in sys.path: sys.path.insert(0, _p)
from legitimate_quantum_engine.solvers.orchestrator import auto_solve
from shared.constants import get_hamiltonian

for pid in ['TFI_chain_4', 'Heisenberg_4', 'MaxCut_triangle']:
    prob = get_hamiltonian(pid)
    out = auto_solve(prob['matrix'], prob['n_qubits'])
    print(f'{pid:18s} energy={out["energy"]:.6f} method={out.get("method","auto")}')
```

**Takeaway.** Good orchestration matches algorithm to problem structure.
