# 08 — Tensor Networks (MPS)

**Concept.** MPS compress area-law states enabling simulation beyond full statevectors.

```python
import os, sys
_LQE = os.environ.get("LQE_PATH", os.path.abspath(os.path.join("..", "Legitimate-Quantum-Engine")))
if _LQE not in sys.path:
    sys.path.insert(0, _LQE)
from legitimate_quantum_engine.tensor_networks.mps_simulator import mps_ghz_state
from legitimate_quantum_engine.core.statevector import StatevectorSim
import numpy as np

n=4
mps = mps_ghz_state(n)
sim = StatevectorSim(n)
sim.apply('H', 0)
for q in range(n-1): sim.apply('CNOT', q, q+1)
print(f'MPS type: {type(mps).__name__}')
print(f'GHZ nonzero: {np.where(np.abs(sim.state)>0.1)[0]}')
```

**Takeaway.** Area-law states stay compressible with MPS.
