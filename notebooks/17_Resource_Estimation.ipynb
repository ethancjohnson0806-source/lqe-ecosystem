# 17 — Resource Estimation

**Concept.** Estimate qubits, depth, and runtime before burning device time.

```python
import os, sys
_LQE = os.environ.get("LQE_PATH", os.path.abspath(os.path.join("..", "legitimate_quantum_engine_v5.0")))
if _LQE not in sys.path: sys.path.insert(0, _LQE)
from legitimate_quantum_engine.benchmarks.resource_estimator import ResourceEstimator

circuit = []
n = 8
for layer in range(3):
    for q in range(n): circuit.append(('ry', q))
    for q in range(n-1): circuit.append(('cx', q, q+1))
est = ResourceEstimator(circuit, hardware='ibm_eagle')
print(est.analyze())
```

**Takeaway.** Resource estimates show when you need better hardware.
