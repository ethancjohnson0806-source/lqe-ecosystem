# 10 — Quantum Chemistry

**Concept.** Molecular Hamiltonians map to qubits; VQE approximates ground-state energies.

```python
import os, sys
_LQE = os.environ.get("LQE_PATH", os.path.abspath(os.path.join("..", "Legitimate-Quantum-Engine")))
_ECO = os.environ.get("LQE_ECOSYSTEM_PATH", os.path.abspath(".."))
for _p in (_LQE, _ECO):
    if _p not in sys.path: sys.path.insert(0, _p)
from quantum_db.api import QuantumDB
from legitimate_quantum_engine.solvers.vqe_ground import VQE
import numpy as np

db = QuantumDB()
H = db.to_matrix('h2_molecule')
exact = float(np.min(np.real(np.linalg.eigvalsh(H))))
out = VQE(2, H, num_layers=2, max_iterations=80).solve(verbose=False)
print(f'Exact: {exact:.6f}  VQE: {out["energy"]:.6f}')
```

**Takeaway.** Chemistry is a leading candidate for early quantum advantage.
