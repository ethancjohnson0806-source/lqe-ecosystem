# 16 — Quantum Volume

**Concept.** Quantum volume (QV) is a single-number benchmark: the largest square random circuit (width = depth = n) that a device (or simulator) can run successfully.

Why it matters: it compresses gate fidelity, connectivity, and compiler quality into one comparable metric.

**Math.**

Success when heavy-output generation probability exceeds 2/3 (HOG criterion). Reported QV is $2^n$ for the largest passing n.

```python
import os, sys
_LQE = os.environ.get("LQE_PATH", os.path.abspath(os.path.join("..", "legitimate_quantum_engine_v5.0")))
if _LQE not in sys.path:
    sys.path.insert(0, _LQE)
from legitimate_quantum_engine.characterization.quantum_volume import QuantumVolume

qv = QuantumVolume(n_qubits=3, n_trials=10, backend="statevector")
report = qv.benchmark()
print("QV benchmark report:")
print(report)
if hasattr(qv, "print_report"):
    qv.print_report()
print("Quantum volume demo finished (ideal simulator passes small n).")
```
